```python
import numpy as np

arr1 = np.array(
    [
        [
            [1, 1, 1],  # Frame 1 – marks in top-left area only
            [0, 0, 0],
            [0, 0, 0],
        ],
        [
            [1, 1, 1],  # Frame 2 – adds a mark at bottom-right
            [0, 0, 0],
            [1, 0, 1],
        ],
        [
            [1, 1, 1],  # Frame 3 – top row full + scattered middle/bottom
            [0, 1, 0],
            [1, 0, 1],
        ],
        [
            [1, 1, 1],  # Frame 4 – nearly fully filled / heavily scribbled
            [1, 1, 1],
            [1, 0, 1],
        ],
    ]
)

# Bottom row: scribbles start middle-left and expand vertically/outward
arr2 = np.array(
    [
        [
            [0, 0, 0],  # Frame 1 – marks only in middle row, left side
            [1, 1, 1],
            [0, 0, 0],
        ],
        [
            [1, 0, 0],  # Frame 2 – expands to top & bottom of left column
            [1, 1, 1],
            [1, 0, 0],
        ],
        [
            [1, 0, 0],  # Frame 3 – more cells filled across all rows
            [1, 1, 1],
            [1, 0, 1],
        ],
        [
            [1, 1, 1],  # Frame 4 – very heavy fill, especially top row
            [1, 1, 1],
            [1, 0, 1],
        ],
    ]
)

arr3 = np.full((4, 3, 3), 0.5)

print(f"MSE of arr1 and arr2: {np.mean((arr1 - arr2) ** 2)}")
print(f"MSE of arr1 and arr3: {np.mean((arr1 - arr3) ** 2)}")
print(f"MSE of arr2 and arr3: {np.mean((arr2 - arr3) ** 2)}")
```
