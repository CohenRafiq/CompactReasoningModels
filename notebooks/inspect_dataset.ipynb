```python
def display_grid(ax, grid, clues=None, title=''):
    ax.imshow(grid.cpu().numpy(), cmap='gray_r', interpolation='nearest')
    ax.set_xlim(-0.5, grid.shape[1] - 0.5)
    ax.set_ylim(grid.shape[0] - 0.5, -0.5)
    for i in range(grid.shape[0] + 1):
        ax.axhline(i - 0.5, color='black', linewidth=1, clip_on=False)
    for j in range(grid.shape[1] + 1):
        ax.axvline(j - 0.5, color='black', linewidth=1, clip_on=False)
    ax.axis('off')
    if clues is not None:
        for i, row_clue in enumerate(clues[0]):
            clue_str = ",".join(str(int(num)) for num in row_clue if num > 0) or "0"
            ax.text(-0.6, i, clue_str, va='center', ha='right', fontsize=12)
        for j, col_clue in enumerate(clues[1]):
            clue_str = ",".join(str(int(num)) for num in col_clue if num > 0) or "0"
            ax.text(j, -0.6, clue_str, va='bottom', ha='center', fontsize=12)
    ax.set_title(title, fontsize=14)
```

```python
def convert_string_data(data_string):
    parsed = ast.literal_eval(data_string)
    result = []
    for item in parsed:
        matrix_str, label = item
        matrix = ast.literal_eval(matrix_str)
        result.append([matrix, label])
    return result
```

```python
for a in range(8):
    data = convert_string_data(df["intermediate_solutions"][a])
    grid_width = 4
    grid_height = (len(data) + grid_width - 1) // grid_width
    fig, axes = plt.subplots(
        grid_height,
        grid_width,
        figsize=(20, 5 * grid_height),
        squeeze=False,          # always returns a 2‑D ndarray
    )
    print(f"Row clues for solution {a}: {df['row_clues'][a]}")
    print(f"Column clues for solution {a}: {df['col_clues'][a]}")
    for i, (matrix, label) in enumerate(data[::-1]):
        row = i // grid_width
        col = i % grid_width
        display_grid(axes[row, col], torch.tensor(matrix), title=label)
    total_cells = grid_height * grid_width
    if len(data) < total_cells:
        for j in range(len(data), total_cells):
            row = j // grid_width
            col = j % grid_width
            axes[row, col].axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
from itertools import zip_longest

def pad_nested_lists(data, max_len=3):
    padded = [
        [inner + [0] * (max_len - len(inner)) for inner in outer]
        for outer in data
    ]
    return np.array(padded, dtype=float)

# Alternative using numpy.pad
def pad_nested_lists_np(data, max_len=3):
    """
    Pad inner lists with zeros using numpy operations.
    """
    # Convert to object array first to handle variable lengths
    result = []
    for outer in data:
        padded_outer = []
        for inner in outer:
            arr = np.array(inner, dtype=float)
            if len(arr) < max_len:
                arr = np.pad(arr, (0, max_len - len(arr)), mode='constant')
            padded_outer.append(arr)
        result.append(np.stack(padded_outer))
    
    return np.stack(result)

# Usage
inputs = [[[3], [1, 1, 1], [1, 1], [1], [1]], [[2], [1], [2, 2], [1], [2]]]

print("Method 1 (list comprehension):")
output1 = pad_nested_lists(inputs, max_len=3)
print(output1)

print("\nMethod 2 (numpy.pad):")
output2 = pad_nested_lists_np(inputs, max_len=3)
print(output2)
```
